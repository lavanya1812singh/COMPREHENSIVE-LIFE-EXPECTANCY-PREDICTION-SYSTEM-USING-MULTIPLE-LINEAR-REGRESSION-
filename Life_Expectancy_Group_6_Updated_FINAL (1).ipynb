# Dealing with Missing Data

```python
df_copy.info()
```

It has 22 columns and nearly 3000 entries and with the inequality in no. of observations from each feature it is clear that it has got some null values. Lets try to fix them up and try to solve the problem.

It has 20 Quantitative and 2 Qualitative features.

```python
df_copy.isna().sum()
```

We can see in columns like 'Life Expectancy' and 'Adult Mortality' there were few NaNs values.
So we have filled respective average value as they are Continous features with very few outliers.

```python
df_copy['Life expectancy']=df_copy['Life expectancy'].fillna(value=df_copy['Life expectancy'].mean())

# For Adult Mortality.

df_copy['Adult Mortality']=df_copy['Adult Mortality'].fillna(value=df_copy['Adult Mortality'].mean())
```

```python
df_copy.isna().sum()
```

There are many columns with null values but the amount of missing values is not big enough to drop the columns. So, to impute the missing values would be a good idea.

Filling missing values with central tendency mean would not be the good idea because of the outliers.

We can also fill it with median but since data is divided into countries by years the median value will vary country wise.
Hence we are imputing with Median values not generally but specific to country wise median.

```python
imputed_data = []
for year in list(df_copy.Year.unique()):
    year_data = df_copy[df_copy.Year == year].copy()
    for col in list(year_data.columns)[4:]:
        year_data[col] = year_data[col].fillna(year_data[col].dropna().median()).copy()
    imputed_data.append(year_data)
df_copy = pd.concat(imputed_data).copy()
```

Capping based off IQR (Interquartile Range datapoints)

```python
Q1 = df_copy[cols].quantile(0.25)
Q3 = df_copy[cols].quantile(0.75)
IQR = Q3 - Q1

outliers = ((df_copy[cols] < (Q1 - 1.7 * IQR)) | (df_copy[cols] > (Q3 + 1.7 * IQR))).any(axis=1)

life_expectancy_data = df_copy[~outliers]

life_expectancy_data
```

# Performance/Evaluation Metrics

 - MAE quantifies the average absolute difference between the actual and predicted values.
 - MSE calculates the average squared difference between the actual and predicted values.
 - RMSE provides an interpretable metric by taking the square root of MSE. It's useful for understanding the error magnitude in the same units as the target variable.

 These metrics allow us to gauge how well the model performs in estimating the target variable. Smaller values of MAE, MSE, and RMSE indicate better model accuracy and a closer fit to the actual data.


```python
#Make predictions on new or unseen data
test_predictions = Linear_model.predict(X_test)
test_predictions
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
# Calculate Mean Absolute Error (MAE)
MAE = mean_absolute_error(y_test, test_predictions)
MAE
```

```python
# Calculate Mean Squared Error (MSE)
MSE = mean_squared_error(y_test, test_predictions)
MSE
```

```python
# Calculate Root Mean Squared Error (RMSE)
RMSE = np.sqrt(MSE)
RMSE
```

```python
# Calculate r2_score
r2_value = r2_score(y_test, test_predictions)
r2_value
```

# Hyperparameter Tuning: Grid Search

Grid Search
A search consists of:

an estimator (regressor or classifier such as sklearn.svm.SVC());                                     
a parameter space;                                                            
a method for searching or sampling candidates;                                                
a cross-validation scheme                                                      
a score function.            

```python
from sklearn.linear_model import ElasticNet
```

```python
base_elastic_model = ElasticNet()
```

```python

param_grid = {
    'alpha': [0.1, 1, 5, 10, 50, 100],  # List of alpha values to search through
    'l1_ratio': [0.1, 0.5, 0.7, 0.9, 0.95, 0.99, 1]  # List of l1_ratio values to search through
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python

grid_model = GridSearchCV(
    estimator=base_elastic_model,
    param_grid=param_grid,  # Dictionary of hyperparameters and their possible values to search through
    scoring='neg_mean_squared_error',
    cv=5,
    verbose=2
)
```

```python
grid_model.fit(X_train,y_train)
```

```python
# Accessing the Best Estimator from Grid Search
best_estimator = grid_model.best_estimator_
```

```python
grid_model.best_params_
```

```python
y_pred = grid_model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(y_test,y_pred)
```
